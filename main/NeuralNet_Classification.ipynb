```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, BatchNormalization, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# ---------------------------
# 1. CLEAN + ALIGN DATA FIRST
# ---------------------------
# Ensure no NaNs in target
df = df.dropna(subset=["Life expectation"])

# Create classes
y_cls = pd.qcut(df["Life expectation"], q=4, labels=False, duplicates='drop')

# Remove rows where qcut failed
mask = ~y_cls.isna()
df = df.loc[mask]
y_cls = y_cls.loc[mask]

# Reset index to avoid alignment issues
df = df.reset_index(drop=True)
y_cls = y_cls.reset_index(drop=True)

# Features
X_cls = df.drop(columns=["Life expectation"])

# ---------------------------
# 2. BUILD MODEL
# ---------------------------
def build_model(input_dim, num_classes):
    m = Sequential([
        Dense(64, input_shape=(input_dim,), activation="relu"),
        BatchNormalization(),
        Dropout(0.3),

        Dense(32, activation="relu"),
        BatchNormalization(),

        Dense(num_classes, activation="softmax")
    ])
    m.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )
    return m

# ---------------------------
# 3. SCALERS
# ---------------------------
scalers = {
    "MinMax":   MinMaxScaler(),
    "Standard": StandardScaler(),
    "Robust":   RobustScaler(),
}

results = {}
es = EarlyStopping(monitor="val_loss", patience=15, restore_best_weights=True)

# ---------------------------
# 4. CLASS WEIGHTS (FIXED)
# ---------------------------
classes = np.unique(y_cls)
weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_cls)
class_weights = dict(zip(classes, weights))

print("Class distribution:", dict(zip(*np.unique(y_cls, return_counts=True))))
print("Class weights:", class_weights)

# ---------------------------
# 5. TRAIN LOOP
# ---------------------------
for name, scaler in scalers.items():

    # Fit scaler ONLY on X
    X_scaled = scaler.fit_transform(X_cls)

    # Split (now safe)
    X_tr, X_te, y_tr, y_te = train_test_split(
        X_scaled, y_cls, test_size=0.15, random_state=42, stratify=y_cls
    )

    tf.random.set_seed(42)

    m = build_model(input_dim=X_tr.shape[1], num_classes=len(classes))

    m.fit(
        X_tr, y_tr,
        epochs=150,
        batch_size=32,
        validation_split=0.15,
        verbose=0,
        callbacks=[es],
        class_weight=class_weights
    )

    _, acc = m.evaluate(X_te, y_te, verbose=0)

    results[name] = {
        "scaler": scaler,
        "model": m,
        "X_test": X_te,
        "y_test": y_te,
        "acc": acc
    }

    print(f"{name:10s}  test accuracy: {acc:.4f}")

# ---------------------------
# 6. BEST MODEL
# ---------------------------
best_name = max(results, key=lambda k: results[k]["acc"])
print(f"\nBest scaler: {best_name} ({results[best_name]['acc']:.4f})")

results[best_name]["model"].summary()
```
