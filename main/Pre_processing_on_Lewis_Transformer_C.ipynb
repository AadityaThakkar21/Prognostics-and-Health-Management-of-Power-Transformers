```python
def hampel_filter_ts(series, window=7, n_sigmas=3, replace_with='median'):
    """
    Hampel filter for a pandas Series indexed by time.
    - window: odd integer number of samples for the rolling window (e.g., 7)
    - n_sigmas: threshold multiple of MAD to call outlier
    - replace_with: 'median' or 'interpolate'
    Returns (corrected_series, outlier_mask)
    """
    s = series.copy().astype(float)
    k = (window - 1) // 2
    # rolling median
    rolling_median = s.rolling(window=window, center=True, min_periods=1).median()
    # absolute deviations from median
    abs_dev = (s - rolling_median).abs()
    # MAD: median of absolute deviations (robust)
    mad = abs_dev.rolling(window=window, center=True, min_periods=1).median()
    # constant to make MAD comparable to std for normal: ~1.4826
    modified_mad = mad * 1.4826
    threshold = n_sigmas * modified_mad
    outlier_mask = abs_dev > threshold
    corrected = s.copy()
    if replace_with == 'median':
        corrected[outlier_mask] = rolling_median[outlier_mask]
    else:  # 'interpolate'
        corrected[outlier_mask] = np.nan
        corrected = corrected.interpolate(method='time', limit_direction='both')
    return corrected, outlier_mask

def apply_hampel_to_df(df, cols=None, window=7, n_sigmas=3):
    if cols is None:
        cols = df.columns
    corrected_df = df.copy().astype(float)
    masks = {}
    for c in cols:
        corrected_df[c], masks[c] = hampel_filter_ts(df[c], window=window, n_sigmas=n_sigmas)
    return corrected_df, masks
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

def detect_outliers_zscore(series, window=25, n_sigmas=3.5):
    """
    Simple rolling-zscore detector: compute rolling mean and std, mark large deviations.
    window: number of samples for stats
    Returns mask
    """
    rolling_mean = series.rolling(window=window, center=True, min_periods=1).mean()
    rolling_std  = series.rolling(window=window, center=True, min_periods=1).std(ddof=0)
    z = (series - rolling_mean).abs() / (rolling_std + 1e-12)
    return z > n_sigmas

def kalman_impute(series, outlier_mask=None, order=(1,0,0), enforce_stationarity=False):
    """
    Fit SARIMAX(order) that handles missing values and use smoothed in-sample predicted_mean
    to replace missing/outlier points.
    - outlier_mask: boolean mask where True = detected outlier (will be set as missing)
    - order: ARMA order for the state-space model (AR component helps smoothing)
    Returns (corrected_series, predicted_mean)
    """
    s = series.copy().astype(float)
    mask = outlier_mask
    if mask is not None:
        s[mask] = np.nan
    # fit SARIMAX to series with missing values; it uses Kalman filter & handles nan
    model = SARIMAX(s, order=order, enforce_stationarity=enforce_stationarity, enforce_invertibility=False)
    res = model.fit(disp=False)
    # in-sample predicted mean (smoothed)
    pred = res.get_prediction(start=0, end=len(s)-1)
    predicted_mean = pred.predicted_mean
    corrected = s.copy()
    # replace nan / masked positions with predicted/smoothed values
    replace_mask = s.isna()
    corrected[replace_mask] = predicted_mean[replace_mask]
    return corrected, predicted_mean, res

def apply_kalman_pipeline(df, cols=None, detect_window=25, detect_thresh=3.5, ar_order=(1,0,0)):
    if cols is None:
        cols = df.columns
    corrected_df = df.copy().astype(float)
    masks = {}
    preds = {}
    models = {}
    for c in cols:
        mask = detect_outliers_zscore(df[c], window=detect_window, n_sigmas=detect_thresh)
        corrected_df[c], preds[c], models[c] = kalman_impute(df[c], outlier_mask=mask, order=ar_order)
        masks[c] = mask
    return corrected_df, masks, preds, models
```

```python
def create_windows(data, window_size, output=1):
    X, y = [], []
    for i in range(len(data) - window_size - output + 1):
        X.append(data[i:i+window_size])
        y.append(data[i+window_size:i+window_size+output])
    return np.array(X), np.array(y)
```

```python
def calculate_non_overlapping_average(scores: np.ndarray, window_size: int = 60) -> np.ndarray:
    """Calculate non-overlapping window averages of scores.
    Args:
        scores: Input array of scores
        window_size: Size of non-overlapping windows
    Returns:
        Array of averaged scores per window
    """
    scores = np.asarray(scores)
    return np.mean(scores[:len(scores) - len(scores) % window_size].reshape(-1, window_size), axis=1)
```

```python
def apply_ewma(hi_scores, alpha=0.2):
    hi_series = pd.Series(hi_scores)
    smoothed_hi = hi_series.ewm(alpha=alpha, adjust=False).mean()
    return smoothed_hi.tolist()
```

```python
def process_health_indices(train_scores: np.ndarray, val_scores: np.ndarray, test_scores: np.ndarray,
                           window_size: int = 60, alpha: float = 0.2):
    """Process health indices with non-overlapping averaging and EWMA smoothing."""
    # Calculate non-overlapping averages
    averages = [
        calculate_non_overlapping_average(scores, window_size)
        for scores in (train_scores, val_scores, test_scores)
    ]

    # Combine and smooth
    all_scores = pd.concat([pd.Series(avg) for avg in averages])
    smoothed = apply_ewma(all_scores, alpha)

    # Split back into train/val/test
    train_len, val_len = len(averages[0]), len(averages[1])
    train_smoothed = smoothed[:train_len]
    val_smoothed = smoothed[train_len:train_len + val_len]
    test_smoothed = smoothed[train_len + val_len:]

    # Calculate threshold
    threshold = max(np.max(train_smoothed), np.max(val_smoothed))
    print("threshold", threshold)

    return train_smoothed, val_smoothed, test_smoothed, threshold
```
